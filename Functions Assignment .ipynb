# Theory Questions:

### 1. What is the difference between a function and a method in Python?

- A **function** is a standalone block of code designed to perform a specific task can be called independently.
- A **method** is a function that is associated with an object (called on an object).

**Example:**
```python
def func():
    print("Hello")

class MyClass:
    def method(self):
        print("Hello from method")

obj = MyClass()
func()             # function call
obj.method()       # method call
```


### 2. Explain the concept of function arguments and parameters in Python.

- **Parameters:** Names in function definition.
- **Arguments:** Values passed when calling the function.

**Example:**
```python
def add(a, b):
    return a + b

print(add(2, 3))  # Outputs 5
```


### 3.   What are the different ways to define and call a function in Python?

- Use `def` keyword to define a function.
- Call function by name with parentheses.
- Supports positional, keyword, default, variable-length arguments, and lambda functions.

**Example:**
```python
def greet(name="World"):
    print("Hello", name)

greet()         # default
greet("Alice")  # positional
greet(name="Bob")   # keyword

# Lambda function example
square = lambda x: x*x
print(square(4))
```


### 4. What is the purpose of the `return` statement in a Python function?

- Ends function execution and sends a value back to the caller.

**Example:**
```python
def get_sum(a, b):
    return a + b

print(get_sum(2, 5))  # 7
```


### 5.  What are iterators in Python and how do they differ from iterables?

- **Iterable:** Object you can loop over (e.g., list, tuple).
- **Iterator:** An object with __iter__() and __next__() to fetch the next value.

**Example:**
```python
nums = [1, 2, 3]  # iterable
iter_nums = iter(nums)  # iterator
print(next(iter_nums))  # Output: 1
```



### 6.  Explain the concept of generators in Python and how they are defined.

- Special functions using `yield` that generate values lazily, one at a time.
- Defined with `def` and contain `yield`.

**Example:**
```python
def count_up(n):
    for i in range(n):
        yield i
for val in count_up(3):
    print(val)
```


### 7. What are the advantages of using generators over regular functions?

- Use less memory because they generate items on the fly instead of storing entire result.
- Suitable for large data or infinite streams.

**Example:**
```python
def gen():
    for i in range(1000000):
        yield i
```


### 8. What is a lambda function in Python and when is it typically used?

- Anonymous, small, single-expression function.
- Used for short throw-away functions, often in `map()`, `filter()`, `sorted()`.

**Example:**
```python
double = lambda x: x*2
print(double(4))
```


### 9.  Explain the purpose and usage of the `map()` function in Python.

- Applies a function to every item in an iterable, producing a map iterator.
- Used to transform data elements without explicit loops.

**Example:**
```python
nums = [1, 2, 3]
squares = list(map(lambda x: x**2, nums))
print(squares)  # [1, 4, 9]
```


### 10.  What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

| Function | Purpose | Returns | Example Usage |
|--------|---------|--------|--------------|
| map() | Applies a function to every item in an iterable | map object (iterator) | `map(f, [1,2])` |
| filter() | Returns items for which function returns True | filter object | `filter(f, [1,2,3])` |
| reduce() | Aggregates items cumulatively into a single value | single value | `reduce(f, [1,2,3])` |

**Example:**
```python
from functools import reduce
nums = [1,2,3,4]
print(list(map(lambda x: x*2, nums)))   # [2, 4, 6, 8]
print(list(filter(lambda x: x%2==0, nums)))   # [2, 4]
print(reduce(lambda x,y: x+y, nums))   # 10
```

````python
# 6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

def powers_of_two(max_exp):
    for exp in range(max_exp + 1):
        yield 2 ** exp

for value in powers_of_two(5):
    print(value)
````
